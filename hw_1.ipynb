```python
# 1. Создайте класс Chocolate (для определения шоколадного батончика), принимающий следующие аргументы при инициализации:
# - цена батончика (тип - float)
# - добавка к шоколаду (необязательный аргумент, тип - str)
# - наличие орехов (тип - bool)
# - количество орехов (тип - int)
# 2 балла

class Chocolate:
    def __init__(self, price: float, nuts: bool, nuts_count: int, additive: str = None):
        """
        Класс для определения шоколадного батончика
        """
        self.price = price
        self.additive = additive
        self.nuts = nuts
        self.nuts_count = nuts_count
```

```python
# 2. При инициализации необходимо проверять, чтобы цена батончика была больше нуля,
# если цена меньше нуля необходимо вызывать исключение ValueError - 2 балла

class Chocolate:
    def __init__(self, price: float, nuts: bool, nuts_count: int, additive: str=None):
        """
        Класс для определения шоколадного батончика
        """
        # Проверка цены
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")

        # Проверка логической согласованности орехов
        if not nuts and nuts_count > 0:
            raise ValueError("Не может быть орехов при nuts=False")
        if nuts and nuts_count <= 0:
            raise ValueError("При nuts=True количество орехов должно быть больше 0")

        self.price=price
        self.additive=additive
        self.nuts=nuts
        self.nuts_count=nuts_count
```

```python
# 3. Реализуйте метод show_my_chocolate, выводящий на печать "Шоколадный батончик, добавка: {ДОБАВКА}"
# в случае наличия добавки, а иначе должна отобразиться следующая фраза: "Обычный шоколадный батончик" - 1 балл

class Chocolate:
    def __init__(self, price: float, nuts: bool, nuts_count: int, additive: str=None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        if not nuts and nuts_count > 0:
            raise ValueError("Не может быть орехов при nuts=False")
        if nuts and nuts_count <= 0:
            raise ValueError("При nuts=True количество орехов должно быть больше 0")

        self.price=price
        self.additive=additive
        self.nuts=nuts
        self.nuts_count=nuts_count

    def show_my_chocolate(self):
        """Выводит информацию о добавке батончика"""
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")
```

```python
# 4. Реализуйте метод count_nuts, выводящий на печать столько * какое количество орехов присутствует в батончике - 1 балл

class Chocolate:
    def __init__(self, price: float, nuts: bool, nuts_count: int, additive: str=None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        if not nuts and nuts_count > 0:
            raise ValueError("Не может быть орехов при nuts=False")
        if nuts and nuts_count <= 0:
            raise ValueError("При nuts=True количество орехов должно быть больше 0")

        self.price=price
        self.additive=additive
        self.nuts=nuts
        self.nuts_count=nuts_count

    def show_my_chocolate(self):
        """Выводит информацию о добавке батончика"""
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")

    def count_nuts(self):
        """Выводит количество орехов в виде звёздочек"""
        if self.nuts:
            print("*" * self.nuts_count)
        else:
            print("Без орехов")
```

```python
# 5. Переопределите метод __str__, который при применении функции print() к объекты класса Chocolate
# должен возвращать строку "Шоколадный батончик, <ЦЕНА> рублей" - 2 балла

class Chocolate:
    def __init__(self, price: float, nuts: bool, nuts_count: int, additive: str = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        if not nuts and nuts_count > 0:
            raise ValueError("Не может быть орехов при nuts=False")
        if nuts and nuts_count <= 0:
            raise ValueError("При nuts=True количество орехов должно быть больше 0")

        self.price = price
        self.additive = additive
        self.nuts = nuts
        self.nuts_count = nuts_count

    def show_my_chocolate(self):
        """Выводит информацию о добавке батончика"""
        if self.additive:
            print(f"Шоколадный батончик, добавка: {self.additive}")
        else:
            print("Обычный шоколадный батончик")

    def count_nuts(self):
        """Выводит количество орехов в виде звёздочек"""
        if self.nuts:
            print("*" * self.nuts_count)
        else:
            print("Без орехов")

    def __str__(self):
        """Возвращает строковое представление батончика"""
        return f"Шоколадный батончик, {self.price} рублей"
```
